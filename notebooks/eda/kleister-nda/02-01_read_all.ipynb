# 02-01 : Read all the Kleister NDA Data

```python
from typing import Any, Dict, List

import pandas as pd
```

## 1. Configuration

```python
data_path = '../../../data'
raw_data_path = f'{data_path}/raw'
input_path = f'{raw_data_path}/kleister-nda'
header_file = f'{input_path}/in-header.tsv'
```

## 2. Load Data

### 2.1 Functions

#### 2.1.1 Read Input Data

```python
def read_input(set_name:str,
               input_path:str,
               headers:List[str],
               keys_column:str='keys') -> pd.DataFrame:
    """
    Read the compressed input file and return a DataFrame.
    
    Parameters
    ----------
    set_name : str
        The name of the dataset [dev-0, test-A, train].
    input_path : str
        The path to the input file.
    headers : List[str]
        The list of headers to be used for the DataFrame.
    keys_column : str
        The name of the column containing keys. Default is 'keys'.
        
    Returns
    -------
    pd.DataFrame
        A DataFrame containing the data from the input file.
    """
    # read the input file
    df_data = pd.read_csv(
        f'{input_path}/{set_name}/in.tsv.xz',
        sep='\t',
        header=None,
        names=headers,
        quoting=3,
        encoding='utf-8'
    )
    
    # convert the keys column to a list
    df_data[f"{keys_column}_list"] = df_data[keys_column].str.split(' ')
    
    # return the data
    return df_data
    
## test the function
df_dev = read_input('dev-0', input_path, ['filename', 'keys', 'text_djvu', 'text_tesseract', 'text_textract', 'text_best'])
display(df_dev)

keys_list = df_dev.iloc[0].keys_list
print(f"Keys list: {keys_list}")

```

### 2.2 Header File

```python
# read tab separated file with pandas
header = list(pd.read_csv(header_file, sep='\t').columns)
print(header)
```

### 2.2 Read Input Files

```python
set_names = ['dev-0', 'test-A', 'train']
input_data = {}

for set_name in set_names:
    input_data[set_name] = read_input(set_name, input_path, header)
    print(f'{set_name} data shape: {input_data[set_name].shape}')

# show the dev data for verification
display(input_data['dev-0'])
```
