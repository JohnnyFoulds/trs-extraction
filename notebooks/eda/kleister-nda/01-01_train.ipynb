# 01-01 : Training Data

Load the Kleister NDA dataset to get a better understanding of the data.

```python
import pandas as pd
```

## 1. Configuration

```python
data_path = '../../../data'
raw_data_path = f'{data_path}/raw'
input_path = f'{raw_data_path}/kleister-nda'

header_file = f'{input_path}/in-header.tsv'
train_file = f'{input_path}/train/in.tsv.xz'
expected_file = f'{input_path}/train/expected.tsv'
expected_original_file = f'{input_path}/train/expected-original.tsv'
```

## 2. Load Data

### 2.1 Header File

```python
# read tab separated file with pandas
header = list(pd.read_csv(header_file, sep='\t').columns)
print(header)
```

### 2.2 Train Data

```python
# read tge in.tsv.xz file with pandas
df_train = pd.read_csv(train_file, sep='\t', compression='xz', header=None, names=header)

# show the data shape
display(df_train)
print('Shape:', df_train.shape)
```

### 2.3 Expected Values

```python
df_expected = pd.read_csv(expected_file, sep='\t', header=None, names=['expected'])

display(df_expected)
print('Shape:', df_expected.shape)
```

```python
df_expected_original = pd.read_csv(expected_original_file, sep='\t', header=None, names=['expected'])

display(df_expected_original)
print('Shape:', df_expected_original.shape)
```

## 3. Preprocessing

### 3.1 Expected Values to Dictionary

The expected values for each row are in the following string format:

`effective_date=2007-01-03 jurisdiction=New_York party=Potomac_Key_Group_LLC party=Compudyne_Corporation term=2_years`

```python
# convert the string to a dictionary
def convert_to_dict(s):
    d = {}
    for item in s.split(' '):
        key, value = item.split('=')
        d[key] = value
    return d

# convert the expected values to a list of dictionaries
df_expected['dict'] = df_expected['expected'].apply(lambda x: convert_to_dict(x))
# show the first 5 rows
display(df_expected.head())
print('Shape:', df_expected.shape)
```
