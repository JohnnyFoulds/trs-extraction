# 01-01 : Nougat OCR

Perform a basic test of OCR on a PDF file using the Nougat OCR engine.

```python
import os
import subprocess
import sys
```

## 1. Configuration

```python
data_path = '../../../data'
test_path = f'{data_path}/test'
samples_path = f'{test_path}/sample_pdf'

output_path = f'{data_path}/output'
output_pdf_path = f'{output_path}/sample_pdf'
current_output_path = f'{output_pdf_path}/nougat'
```

## 2. Get Sample PDF List

```python
# get a list of all the pdf files in the sample_pdf directory
pdf_files = os.listdir(samples_path)
pdf_files = [f for f in pdf_files if f.endswith('.pdf')]
pdf_files.sort()

# show the list of pdf files
print('List of PDF files:')
print(pdf_files)
```

## 3. Basic Test

```python
# get the absolute path the the pdf file
pdf_file = os.path.abspath(f'{samples_path}/{pdf_files[4]}')
print(f'Absolute path to the first PDF file: {pdf_file}')

# call nougat
!nougat {pdf_file} -o {current_output_path} -m 0.1.0-base --recompute
```

## 4. Process All PDF Files

```python
for file in pdf_files:
    # get the absolute path to the pdf file
    pdf_file = os.path.abspath(f'{samples_path}/{file}')
    print(f'Absolute path to the PDF file: {pdf_file}')

    # call nougat
    subprocess.run(['nougat', pdf_file, '-o', current_output_path, '-m', '0.1.0-base', '--recompute'])
```
