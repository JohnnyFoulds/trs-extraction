```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from mistralai import Mistral
import os
from IPython.display import display, Markdown
```

```python
data_path = "../../../data"
input_path = f"{data_path}/papers" 
output_path = f"{data_path}/output/papers"

file_name = "02_rag_qa_design"

paper_file = f"{input_path}/{file_name}.pdf"
output_file = f"{output_path}/{file_name}.md"
```

```python
client = Mistral(api_key=os.getenv("MISTRAL_API_KEY"))
```

```python
uploaded_pdf = client.files.upload(
    file={
        "file_name": "uploaded_file.pdf",
        "content": open(paper_file, "rb"),
    },
    purpose="ocr"
)  
```

```python
signed_url = client.files.get_signed_url(file_id=uploaded_pdf.id)
```

```python
ocr_response = client.ocr.process(
    model="mistral-ocr-latest",
    document={
        "type": "document_url",
        "document_url": signed_url.url,
    },
    include_image_base64=False
)
```

```python
#markdown_str = "\n\n".join([f"### Page {i+1}\n{ocr_response.pages[i].markdown}" for i in range(len(ocr_response.pages))])
markdown_str = "\n\n".join([ocr_response.pages[i].markdown for i in range(len(ocr_response.pages))])

# save the markdown content to a file
with open(output_file, "w") as f:
    f.write(markdown_str)
```

```python
display(Markdown(markdown_str))
```
